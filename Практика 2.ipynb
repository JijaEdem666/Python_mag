```python
#Задание 1
def print_addition_table():
    for i in range(1, 10):
        for j in range(1, 10):
            print(f"{i}+{j}={i + j}", end="\t")
        print()

def print_subtraction_table():
    for i in range(1, 10):
        for j in range(1, 10):
            print(f"{i}-{j}={i - j}", end="\t")
        print()

def print_multiplication_table():
    for i in range(1, 10):
        for j in range(1, 10):
            print(f"{i}*{j}={i * j}", end="\t")
        print()

def print_division_table():
    for i in range(1, 10):
        for j in range(1, 10):
            print(f"{i}/{j}={i / j:.2f}", end="\t")
        print()


print("Выберите таблицу:")
print("1. Сложение")
print("2. Вычитание")
print("3. Умножение")
print("4. Деление")

choice = input()

if choice == '1':
    print_addition_table()
elif choice == '2':
    print_subtraction_table()
elif choice == '3':
    print_multiplication_table()
elif choice == '4':
    print_division_table()
else:
    print("Ошибка: некорректный ввод. Выбериет выриант от 1 до 4.")
```

```python
#Задание 2
import random

def create_vector(n):
    return [random.random() for _ in range(n)]

def create_matrix(m, n):
    return [[random.random() for _ in range(n)] for _ in range(m)]

def multiply_matrix_vector(matrix, vector):
    m = len(matrix)
    n = len(matrix[0])
    result = [0.0] * m
    for i in range(m):
        for j in range(n):
            result[i] += matrix[i][j] * vector[j]
    return result

def print_matrix(matrix):
    for row in matrix:
        print(" ".join(f"{val:.4f}" for val in row))

def print_vector(vector):
    print(" ".join(f"{val:.4f}" for val in vector))

def sum_diagonal(matrix):
    rows = len(matrix)
    cols = len(matrix[0])
    diag_sum = 0.0
    for i in range(min(rows, cols)):
        diag_sum += matrix[i][i]
    return diag_sum

def convolve2d(image, kernel):
    img_h = len(image)
    img_w = len(image[0])
    ker_h = len(kernel)
    ker_w = len(kernel[0])

    out_h = img_h - ker_h + 1
    out_w = img_w - ker_w + 1
        
    result = [[0.0 for _ in range(out_w)] for _ in range(out_h)]
    
    for i in range(out_h):
        for j in range(out_w):
            conv_sum = 0.0
            for ki in range(ker_h):
                for kj in range(ker_w):
                    conv_sum += image[i + ki][j + kj] * kernel[ki][kj]
            result[i][j] = conv_sum
            
    return result


vec = create_vector(5)

mat = create_matrix(3, 5)

print_vector(vec)

print()

print_matrix(mat)

print()

diag_sum = sum_diagonal(mat)
print(diag_sum)

print()

result_vec = multiply_matrix_vector(mat, vec)

print_vector(result_vec)

print()

image = create_matrix(5, 5)

kernel = create_matrix(3, 3)


print_matrix(image)

print()

print_matrix(kernel)

print()

convolved_image = convolve2d(image, kernel)

print_matrix(convolved_image)
```
