```python
#Задание 1
while True:
    a, b = map(float, input("Введите 2 числа через пробел: ").split())
    print("Сумма введённых чисел: ", a + b)
```

```python
#Задание 2
num = 1
for i in range(1, 6):
    for j in range(1, 6):
        if (i + j) % 2 == 0:
            print("*", end="\t")
        else:
            print(num, end="\t")
            num += 1
    print()
```

```python
#Задание 4
a = int(input())
maxa = a
while a > 0:
    a = int(input())
    if a > maxa:
        maxa = a
print("Максимальное число", maxa)
```
